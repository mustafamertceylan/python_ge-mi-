```python
# fonksiyonlarda dikkat edilmesi gereken bence en önemli mevzu girilen parametrelerin uygunluğudur.
# girilen metindeki bütün karakterleri küçük harfe çeviren metot (upper() ve lower() hazır metotlarını kullanmadan)
def kucuk_harfe_cevir(metin):
    j=0
    list_kucuk=""
    j=0
    for i in metin:
        if 65<=ord(i)<=90:
            list_kucuk+=(chr(ord(i)+32))
            j+=1
            continue
        list_kucuk+=i
        j+=1
    print(list_kucuk)

kucuk_harfe_cevir("Bu BiR OrneK MetinDir...")
        
        
```

```python
# fonksiyonlardaki bir diğer güzel olay ise buradaki girmek zorunda olduğumuz parametrelerin girilmemesi durumu
#eğer eksik girdiysek zaten bize hata veriyor bunu biliyoruz peki hata vermemesi için ne yapmalı....

def hi(metin,isim="default"):
    print(f"{metin} {isim}")

hi("nasılsın","MMC")
hi("nasılsın")
# burada görüldüğü üzere fonksiyona 2 parametre girmemiz gerektiğği halde tek parametre ile hata almadan fonksiyonu çalıştırabildik
```

```python
#return kelimesini kullanarak geriye değer döndürrülmesini yapabilir
def topla(x,y):
    return x+y

print(topla(5,5))
```

```python
#fonksiyonlarda isim değişikliği yapmak mümkündür. Bu işlem aslında değer atamasına çok benziyor. Zaten burada yaptığımız işlem aslında kendi değişkenimizin fonksiyonun adresini göstererek çalışmasıdır
my_sum=topla
print(my_sum(8,5))
```
