```python
import sys
sayi=5
harf="a"
isim="isim"
kusuratli=3.1415
print(sys.getsizeof(sayi))
print(sys.getsizeof(isim))
print(sys.getsizeof(kusuratli))
print(sys.getsizeof(harf))

# C dilindeki gibi bir integer değeri sabit 4 byte veya bir float değer 8 byte gibi sabit bir genişlik mevcut değil
#Bunun sebebi pythonda tutulan her bir değişkenin aslında bir nesne olmasından kaynaklanır (BOXED OBJECTS)
```
