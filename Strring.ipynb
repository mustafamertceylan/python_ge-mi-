```python
files=["dosya1.pdf","resim.jpg","readme.txt","file.pdf"]

pdfs=[]
for file in files:
    newFile=file.split(sep=".")
    if newFile[1]=="pdf":
        pdfs.append(newFile)
print(pdfs)
```

```python
pdfs=[]
for file in files:
    if file.endswith(".pdf"):
        pdfs.append(file)

print(pdfs)
```
