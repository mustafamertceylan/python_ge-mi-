```python
#-------------------------------------------Kümeler (sets)------------------------------#
```

```python
set1={"item1","item2","item3","item4","item5","item6"}
#listelerde yapılan her türlü işlem set veri tipinde de yapılabilmektedir.
#listelerden ve tuple veri türünden ayrılan en belirgin özelliği ise verilerin belli bir sıra ile tutulmamasıdır...
```

```python
for item in set1:
    print(item)

#çıktıda da görüldüğü üzere ekrana yazdırılma sırası birbirinden farklı ve sıralı olarak işlenmedi
#Normalde her seferinde farklı bir sıra ile çıktı almamız gerekir lakin jupyter notebook yapısı gereğği aynı çıktıları üretiyo kafa karışıklığı olmasın
#eğer çekirdeği yeniden başlatıp çalıştırırsak farklı bir sıra ile geleceği aşikar olarak görünebilir😒😒😒😒 
```

```python
print(set1)
set1.add("new")
print(set1)

set1.remove("new")
print(set1)

print(set1.clear())

set1.discard("newitem")  
#hata almamamızın sebebi discard() metodunun yapısı ile alakalıdır...
```

```python
#kümelerden söz ederken en çok kullanılan kesişim ve birleşim metodları bu veri tipinde de mevcuttur...
kume1={"item1","item2","item3","item4"}
kume2={"item1","item5", "item2"}

print(kume1.intersection(kume2))
print(kume1.difference(kume2))
print(kume1.union(kume2))
```

```python
#-------------------------------------------DİCTİONARY---------------------------------------------------
```

```python
#Key  : Value olarak verileri depolamaya yarayan bir türdür.
#Key = Sadece string ve integer türünden geğerler alabilir.
# Value = İstenilen herhangi türde bir değer alabilir...
user_info={"isim" : "Hax","yas" : 25,"hobiler" : ["yüzmek","koşmak","kitap okumak"]}
print(user_info)

print(len(user_info))

print(type(user_info))

print(user_info["hobiler"])
print(user_info["hobiler"][1])
```

```python
print(user_info)
user_info["isim"]="karen"

print(user_info)

#listelerde olduğu gibi sözlük yapısında da value değerlerine müdahale etmek mümkündür...

#eğer birden fazla alana müdahale etmek istersek update() metodunu kullanabiliriz

user_info.update({"isim":"Hans","yas": 45})
print(user_info)
```

```python
#yeni bir key : value ikilisi eklemek istersek eğer 
user_info["id"]= 6955

print(user_info)
```

```python
#key : value ikililerinden birini silmek istersek eğer 
del user_info["id"]
print(user_info)
```

```python
#döngüler ile nasıl yazdırabiliriz🤨🤨🤨🤨🤨🤨🤨🤨

for x in user_info:
    print(x ," : ",user_info[x])
```

```python
print(user_info.keys())
print(user_info.values())
print(user_info.items())
#farklı ve değişik metotlar var hepsine vakit ayıramam ama metot tanımlama kısmında her bir metot için kendi metodumu tanımlamayı planlıyorum
```

```python
print(user_info.get("id"))#hata almadan öyle bir alan olmadığını bize gösterebiliyor

print(user_info["id"])
#olmayan bir değeri görğntülemeye çalışırsak alacağımız hata

```
